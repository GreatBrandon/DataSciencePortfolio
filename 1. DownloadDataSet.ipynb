```python
import kagglehub
import shutil
import os

datasets = ["ronakbadhe/chess-evaluations", "sergionefedov/chess-rating-plateau-why-players-get-stuck"]
cwd = os.getcwd()

for dataset in datasets:
    # Download dataset
    path = kagglehub.dataset_download(dataset)

    # Copy downloaded files into the current directory
    for file in os.listdir(path):
        src = os.path.join(path, file)
        dst = os.path.join(cwd, file)

        if os.path.isfile(src):
            shutil.copy2(src, dst)

    print("Dataset files copied to:", cwd)
```
